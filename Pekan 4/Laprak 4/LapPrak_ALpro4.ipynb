## 2.1 Dasar Teori

1. If Tunggal: Digunakan saat program hanya perlu mengeksekusi aksi jika suatu kondisi terpenuhi, dan mengabaikannya jika kondisi bernilai False.

2. If-Else: Struktur percabangan dua jalur. Jika kondisi bernilai True, blok if dieksekusi; jika False, blok else yang dijalankan.

3. If-Elif-Else: Struktur percabangan multi-jalur mutlak (mutually exclusive). Evaluasi dilakukan secara berurutan dari atas ke bawah. Ketika salah satu kondisi bernilai True, blok kode tersebut dieksekusi dan seluruh sisa blok elif/else di bawahnya diabaikan.

4. Multi-IF Terpisah: Struktur yang menggunakan beberapa blok if mandiri secara berurutan. Setiap blok if dievaluasi secara independen sehingga memungkinkan pengakumulasian nilai (seperti akumulasi diskon) apabila lebih dari satu kondisi terpenuhi secara bersamaan.

5. Match-Case: Fitur structural pattern matching yang diperkenalkan pada Python 3.10. Digunakan untuk menggantikan rantai if-elif panjang pada pemilihan nilai diskrit, di mana case _ berfungsi sebagai wildcard (pilihan default).

### 2.2.1 Program IF 1 (if1_2611533013.py)


```python
# Buat file dengan nama if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input() 

ipk_3013 = float(input("Input IPK Anda: "))

if ipk_3013 >= 2.75:
    print("Anda Lulus Sangat Memuaskan dengan ipk " + str(ipk_3013))

print("Program Selesai")
```

    Program mengevaluasi apakah `ipk_3013` lebih besar dari 2.75. Jika bernilai `True`, pesan kelulusan dicetak. Jika `False`, blok `if` dilewati.

### 2.2.2 Program IF 2 (if2_2611533013.py)

    `else:` <br>
    Jika kondisi `ipk_3013 > 2.75` bernilai `True`, blok `if` dieksekusi. Jika bernilai `False`    (IPK $\leq$ 2.75), kontrol dialihkan ke blok `else` untuk mencetak \"Anda Tidak Lulus\".
    

### 2.2.3 Program  Multi-IF 1 (multi_if1_2611533013.py)

```python
# Buat file dengan nama multi_if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input() 

umur_3013 = int(input("Input umur Anda: "))
sim_3013 = input("Apakah Anda Sudah Punya Sim C (y/t): ")[0]

if umur_3013 >= 17 and sim_3013 == 'y':
    print("Anda  Sudah dewasa dan boleh bawa motor")

if umur_3013 < 17 and sim_3013 != 'y':
    print("Anda Sudah dewasa tetapi tidak boleh bawa motor")

if umur_3013 < 17 and sim_3013 == 'y':
    print("Anda Belum Cukup Umur punya SIM")

if umur_3013 > 17 and sim_3013 != 'y':
    print("Anda Belum Cukup Umur bawa motor")
```

### 2.2.4 Program IF-ELIF-ELSE 1 (if_elif_else1_2611533013.py)

### 2.2.5 Program Multi-IF 2 (multi_if2_2611533013.py)

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_3013 = int(input("Input Total Belanja Anda = Masukkan total belanja (RP): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_3013 = input("Apakah Anda Member (y/t): ").strip().lower()
is_member_3013 = input_member_3013 in ["y", "ya"]

# Input status promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3013 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_Valid_3013 = input_promo_3013 in ['y', 'ya']

total_diskon_persen_3013 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3013 > 100000:
    total_diskon_persen_3013 += 10  # Diskon belanja besar

if is_member_3013:
    total_diskon_persen_3013 += 5  # Diskon member

if kode_promo_Valid_3013:
    total_diskon_persen_3013 += 15  # Diskon voucher

# menghitung nominal diskon dan total bayar
nominal_diskon_3013 = total_belanja_3013 * (total_diskon_persen_3013 / 100)
total_bayar_3013 = total_belanja_3013 - nominal_diskon_3013

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon: {total_diskon_persen_3013}% (Rp {nominal_diskon_3013:,.0f})")
print(f"Total Bayar: Rp {total_bayar_3013:,.0f}")

print (f"Total diskon yang anda dapatkan adalah {total_diskon_persen_3013}%")
# Output: Total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan**:

1. String Handling dan Pengecekan Keanggotaan List<br>
is_member_3013 = input_member_3013 in ["y", "ya"]<br>
Menggunakan .strip().lower() untuk membersihkan spasi dan mengonversi masukan ke huruf kecil, lalu mengecek apakah input berada dalam list ["y", "ya"].<br>

2. Penumpukan Diskon Akumulatif<br>
if total_belanja_3013 > 100000:<br>
 total_diskon_persen_3013 += 10<br>
if is_member_3013:<br>
 total_diskon_persen_3013 += 5<br>
if kode_promo_Valid_3013:<br>
 total_diskon_persen_3013 += 15<br>
Karena menggunakan blok if terpisah, variabel total_diskon_persen_3013 akan ditambahkan nilainya secara bertahap sesuai dengan setiap syarat yang terpenuhi.

### 2.2.7 Program Match Case 1 (match_case1_2611533013.py)

```python
# Buat file dengan nama match_case1.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input() 
# Program konversi angka menjadi nama bulan

bulan_3013 = int(input("Masukkan angka bulan (1-12): "))

match bulan_3013:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan**:


1. Pemadanan Pola dengan Match-Case<br>
match bulan_3013:<br>
Mencocokkan nilai variabel bulan_3013 dengan deretan opsi case.

2. Wildcard Case Default<br>
case _:<br>
Simbol _ bertindak sebagai penampung default jika input angka pengguna berada di luar rentang 1–12.
